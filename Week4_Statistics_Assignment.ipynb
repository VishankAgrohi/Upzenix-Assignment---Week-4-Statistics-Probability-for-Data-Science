# Week 4 Assignment: Statistics & Probability for Data Science

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

# Load dataset
df = pd.read_csv('dataset_study.csv')

# Display first five rows
df.head()
```

## Dataset Information

```python
# Dataset shape and summary
print(df.shape)
print(df.describe())
```

## Correlation Analysis

```python
# Compute correlation
correlation = df['study_hours'].corr(df['grade'])
print("Correlation between study hours and grades:", correlation)
```

## Scatter Plot

```python
plt.figure(figsize=(6,4))
plt.scatter(df['study_hours'], df['grade'], alpha=0.6)
plt.xlabel('Study Hours')
plt.ylabel('Exam Scores')
plt.title('Study Hours vs Exam Scores')
plt.grid(True)
plt.show()
```

## Histogram

```python
plt.figure(figsize=(6,4))
plt.hist(df['study_hours'], bins=20)
plt.xlabel('Study Hours')
plt.ylabel('Frequency')
plt.title('Distribution of Study Hours')
plt.grid(True)
plt.show()
```

## Hypothesis Testing

```python
# Hypothesis Test using Linear Regression
slope, intercept, r_value, p_value, std_err = stats.linregress(df['study_hours'], df['grade'])

print("Slope:", slope)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis: Study hours significantly affect exam scores.")
else:
    print("Fail to reject the null hypothesis: Study hours do not significantly affect exam scores.")
```


## Conclusion

- The correlation coefficient between study hours and exam scores is **0.8263**, indicating a strong positive relationship.
- The p-value obtained from hypothesis testing is **0.000000**.
- Since the p-value is less than 0.05, the null hypothesis is rejected.
- Therefore, study time has a statistically significant effect on exam performance.
